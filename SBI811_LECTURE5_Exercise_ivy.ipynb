# DNA Toolbox - Function Continuation

**Functions implemented:**
1. `clean_sequence()` - base cleaning function
2. `is_valid_dna()` - calls clean_sequence() then validates
3. `calculate_gc_percentage()` - calls clean_sequence() then calculates GC%
4. `get_length()` - MY OWN FUNCTION - calls clean_sequence() then returns len()

**Continuation principle:** All functions reuse clean_sequence() instead of rewriting cleaning logic.

**Bug Fix Report:**
- In calculate_gc_percentage line 24, original code was `gc_count == 1`
- This was a comparison, not an addition, so gc_count stayed 0
- Fixed to `gc_count += 1` to correctly increment count

All functions tested and working.

```python
def clean_sequence(sequence):
    """"Clean a DNA sequence by stripping whitespace and converting to uppercase."""
    return sequence.strip().upper()

def is_valid_dna(sequence):
    """"checks if seuence contains valid bases."""
    cleaned_sequence = clean_sequence(sequence)
    if len(cleaned_sequence) == 0:
        return False
    for base in cleaned_sequence:
        if base not in "ATGC":
            return False
    return True
        
def calculate_gc_percentage(sequence):
    """"clalculate the GCpercentage."""
    
    cleaned_sequence = clean_sequence(sequence)
    if len(cleaned_sequence) == 0:
        return 0.0
    gc_count = 0
    for base in cleaned_sequence:
        if base == "G" or base == "C":
            gc_count += 1

    gc_percent = (gc_count/len(cleaned_sequence)) * 100
    return gc_percent

def get_length(sequence):
    """" calculate the length of the sequence."""
    cleaned_sequence = clean_sequence(sequence)
    length = len(cleaned_sequence)
    return length
```

```python
for seq in [" ATGCATGC ", "GGCC", "ATGCNN", ""]:
    print(f"Seq: '{seq}' -> Clean: {clean_sequence(seq)}, Len: {get_length(seq)}, Valid: {is_valid_dna(seq)}, GC%: {calculate_gc_percentage(seq):.1f}%")
```
